```python
# Correct gender column
df['sex'] = df['sex'].map({1: 'Male', 0: 'Female'})
```

```python
# correct exange column
df['exang'] = df['exang'].map({1: 'Yes', 0: 'No'})
```

```python
# Correct fbs column
df['fbs'] = df['fbs'].map({1: 'High', 0: 'Normal'})
```

```python
df.columns = df.columns.str.strip()
```

```python
df.describe()
```

```python
# Kitni rows affected hain check 
print((df['ca'] == 4).sum())
print((df['thal'] == 0).sum())
```

```python
# Drop
df = df[df['ca'] != 4]
df = df[df['thal'] != 0]
```

# Connect to MS SQL Server

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# SQL Server connection
host = r"DESKTOP-N9K70SP\SQLEXPRESS"
database = "HeartDiseaseDB"

# ODBC Driver
driver = quote_plus("ODBC Driver 17 for SQL Server")

# Create SQL Server connection
engine = create_engine(
    f"mssql+pyodbc://@{host}/{database}"
    f"?driver={driver}&trusted_connection=yes"
)

# Write DataFrame to SQL Server
df.to_sql(
    "heart",
    engine,
    if_exists="replace",
    index=False
)

# Read back sample
result = pd.read_sql(
    "SELECT TOP 5 * FROM heart;", engine )

result 
```
